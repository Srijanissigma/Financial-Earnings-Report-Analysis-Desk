# Financial Earnings Report Analysis Desk
Multi-agent system (Semantic Kernel + Groq): Extraction -> Revenue / Risk / Outlook analysts in parallel -> Editor Manager.

Run `setup_check.py` first. Your key goes in `.env`.

```python
import pathlib
from desk.editor_manager import EditorManager
from desk.config import GROQ_API_KEY, PREFERRED_MODELS
print('Groq key found:', bool(GROQ_API_KEY), '| models:', PREFERRED_MODELS)
```

## 1. Load the report

```python
report = pathlib.Path('data/sample_report.txt').read_text(encoding='utf-8')
print(report)
```

## 2. Run the whole pipeline (fan-out / fan-in)

```python
desk = EditorManager()
note = await desk.run(report)
print(note)
```

## 3. Inspect the shared fact store (every figure with its source sentence)

```python
for f in desk.store.facts:
    print(f'{f.id:>4}  {f.label:<55} {f.display():<12} | {f.source[:70]}')
```

## 4. Each analyst's verdict and which mode it ran in (llm or rules)

```python
for a in desk.analyses:
    print(a.section, '->', a.sentiment, '|', a.verdict, '| mode:', a.mode, '| claims:', len(a.claims), '| dropped:', a.dropped)
```

## 5. Save the note

```python
pathlib.Path('analyst_note.md').write_text(note, encoding='utf-8')
print('saved analyst_note.md')
```
